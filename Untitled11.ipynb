```python
def closest_color(rgb_color):
  color_names = {
      'Red': (255, 0, 0),
      'Green': (0, 255, 0),
      'Blue': (0, 0, 25),
      'Yellow': (255, 255, 0),
      'Purple': (128, 0, 128),
      'Orange': (255, 165, 0),
      'Pink': (255, 192, 203),
      'Brown': (165, 42, 42),
      'Black': (0, 0, 0),
      'White': (255, 255, 255)
  }

  min_distance = float('inf')
  closest_name = None

  for name, color in color_names.items():
      distance = np.linalg.norm(np.array(rgb_color) - np.array(color))
      if distance < min_distance:
          min_distance = distance
          closest_name = name

  return closest_name
```
