```python
def process_image(input_part,output_part):
  img = mpimg.imread(input_part)
  plt.imshow(img)
  plt.axis('off')
  plt.show()
  grayscale_image = cv2.cvtColor(img,cv2.COLOR_RGB2GRAY)
  cv2.imwrite(output_part,grayscale_image)
  print(f'saved {output_part}')
  grayscale_image = mpimg.imread(output_part)
  plt.imshow(grayscale_image,cmap='gray')
  plt.axis('off')
  plt.show()
```

```python
input_image_path = '/content/Untitled Project - .png'
output_image_path = '/content/Untitled Project - gray.png'
process_image(input_image_path,output_image_path)
```

```python
# prompt: buatkan code yang mengganti warna img menjadi hitam

import cv2
import matplotlib.pyplot as plt

# Load the image
img = cv2.imread('/content/drive/MyDrive/Colab Notebooks/rumah-sederhana.jpg')

# Convert the image to grayscale
gray_image = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)

# Invert the grayscale image to get a "negative"
inverted_image = 255 - gray_image

# Apply a colormap to the inverted grayscale image to simulate black color
colored_image = cv2.applyColorMap(inverted_image, cv2.COLORMAP_BONE)


# Display the original and the modified images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))  # Convert BGR to RGB for display
plt.title('Original Image')

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(colored_image, cv2.COLOR_BGR2RGB))  # Convert BGR to RGB for display
plt.title('Blackened Image')

plt.show()

# Save the blackened image (optional)
cv2.imwrite('blackened_image.jpg', colored_image)
```

```python
# prompt: buatkan saya code yang bi mengubah warana variable img menjadi putih

# Load the image
img = cv2.imread('/content/drive/MyDrive/Colab Notebooks/rumah-sederhana.jpg')

# Convert the image to white
white_image = np.full(img.shape, 255, dtype=np.uint8)

# Display the original and the modified images
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))  # Convert BGR to RGB for display
plt.title('Original Image')

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(white_image, cv2.COLOR_BGR2RGB))  # Convert BGR to RGB for display
plt.title('White Image')

plt.show()

# Save the white image (optional)
cv2.imwrite('white_image.jpg', white_image)
```
